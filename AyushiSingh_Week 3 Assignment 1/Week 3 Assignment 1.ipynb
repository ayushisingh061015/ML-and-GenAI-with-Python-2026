```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv("agriculture_yield_dataset.csv")

print("Dataset Loaded Successfully!")
print(df.shape)
df.head()
```

```python
print(df.columns)
```

```python
#Rainfall vs Yield
plt.figure(figsize=(6,4))
plt.scatter(df["rainfall_mm"], df["yield_ton_per_hectare"])
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (ton/hectare)")
plt.title("Rainfall vs Yield")
plt.grid(True)
plt.show()
```

```python
#Fertilizer vs Yield
plt.figure(figsize=(6,4))
plt.scatter(df["fertilizer_kg"], df["yield_ton_per_hectare"])
plt.xlabel("Fertilizer")
plt.ylabel("Yield")
plt.title("Fertilizer vs Yield")
plt.grid()
plt.show()
```

```python
corr = df.select_dtypes(include="number").corr()

corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
print(corr["yield_ton_per_hectare"].sort_values(ascending=False))
```

```python
crop_avg = df.groupby("crop_type")["yield_ton_per_hectare"].mean()

crop_avg
```

```python
soil_avg = df.groupby("soil_type")["yield_ton_per_hectare"].mean()

soil_avg
```

```python
print("Highest Average Yield Crop:", crop_avg.idxmax())
print("Highest Average Yield Soil:", soil_avg.idxmax())
```

```python
categorical = df.select_dtypes(include="object").columns

print(categorical)
```

```python
encoded_df = pd.get_dummies(df, columns=categorical)

encoded_df.head()
```

```python
X = encoded_df.drop("yield_ton_per_hectare", axis=1)

y = encoded_df["yield_ton_per_hectare"]

print("Input Features")
X.head()
```

```python
print("Target Variable")
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
print("Intercept:")
print(model.intercept_)
coef = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coef
print("Feature with Highest Positive Coefficient:")
coef.sort_values(by="Coefficient", ascending=False).head(1)
```
